```python
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

np.random.seed(42)
np.random.seed(42)
```

```python
n = 3000

temperature = np.random.uniform(20,60,n)
current = np.random.uniform(100,250,n)
voltage = np.random.uniform(3.2,5.2,n)
soc = np.random.uniform(10,100,n)
resistance = np.random.uniform(.01,.15,n)
cycles = np.random.uniform(10,1500,n)

X = np.column_stack((

        temperature,
        current,
        voltage,
        soc,
        resistance,
        cycles

))

y = (
    (temperature > 20) |
    (resistance >.08) |
    (voltage < 3.6)
).astype(int)

print("Input shape:",X.shape)
print("Output shape:",y.shape)
print("First five inputs:\n",X[:5])
print("First five outputs:",y[:5])


```

```python
indices  = np.random.permutation(len(X))

X = X[indices]
y = y[indices]


split = int(0.8 * len(X))

X_train = X[:split]
y_train = y[:split]

X_test = X[split:]
y_test = y[split:]

print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)
print("X_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)
```

```python
mean = X_train.mean(axis=0)
std = X_train.std(axis=0)

X_train = (X_train - mean) / std
X_test = (X_test - mean) / std

print(X_train.shape)
print(X_test.shape)
```

```python
model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(6,)),
    tf.keras.layers.Dense(16, activation='relu'),
    tf.keras.layers.Dense(8, activation='relu'),
    tf.keras.layers.Dense(1,'sigmoid')
]

)

model.summary()
```

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
early_stopping = tf.keras.callbacks.EarlyStopping(
    patience=5,
    restore_best_weights=True
)

history = model.fit(
    X_train,
    y_train,
    validation_split=0.2,
    batch_size=32,
    epochs=100,
    callbacks = [early_stopping]
)
```

```python
loss,accuracy = model.evaluate(
    X_test,
    y_test

)

print("Test Loss:",loss)
print("Accuracy:",accuracy)
```

```python
plt.plot(history.history['loss'],
         label='Training Loss')

plt.plot(history.history['val_loss'],
         label='Validation Loss')

plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.legend()
plt.show()
```

```python
new_battery = np.array([[
    35,3.8,100,200,80,.04
]])

new_battery = (new_battery - mean) / std

prediction = model.predict(new_battery)

if prediction [0][0]> 0.5:
    print("Warning")
else:
    print("Normal")

```
